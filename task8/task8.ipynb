```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth
```

```python
df_metadata = pd.read_csv('movies_metadata_clean.csv', encoding='iso-8859-1')
df_ratings = pd.read_csv('movies_ratings_small.csv', encoding='iso-8859-1')

# drop unimportant data
df_metadata = df_metadata.drop(["homepage", "imdb_id", "belongs_to_collection", "tagline", "poster_path", "vote_count"], axis=1)
df_metadata.dropna(axis=0, inplace=True)
df_ratings.drop('timestamp', axis = 1)

# keep high ratings
df_ratings_high = df_ratings[df_ratings["rating"] >= 3.5]
```

```python
# create pivot table
pivot = df_ratings_high.pivot(index="userId", columns="movieId", values="rating")
pivot = pivot.notna().astype(int)
pivot
```

```python
frequent_movies1 = apriori(pivot, min_support=0.07, use_colnames=True)
frequent_movies2 = fpgrowth(pivot, min_support=0.07, use_colnames=True)

rules1 = association_rules(frequent_movies1, metric="lift", min_threshold=1.2)
rules2 = association_rules(frequent_movies2, metric="lift", min_threshold=1.2)
```

```python
rules1.sort_values(by=['lift'], ascending=False, inplace=True)
rules1.head()
```

```python
rules2.sort_values(by=['confidence'], ascending=False, inplace=True)
rules2.head()
```
